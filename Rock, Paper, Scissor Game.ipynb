# Task5
### Loop 3 times 

```python
import random

user_score = 0
computer_score = 0

for x in range(3):
    game_options = ['Rock', 'Paper', 'Scissor']
    user_selection = input('Enter your selection Rock, Paper or Scissor: ').capitalize()
    game_options.remove(user_selection)
    computer_selection = random.choice(game_options)

    if user_selection == 'Rock':
        if computer_selection == 'Paper':
            print('Computer win')
            computer_score += 1
        elif computer_selection == 'Scissor':
            print('User win')
            user_score += 1

    elif user_selection == 'Paper':
        if computer_selection == 'Scissor':
            print('Computer win')
            computer_score += 1
        elif computer_selection == 'Rock':
            print('User win')
            user_score += 1

    elif user_selection == 'Scissor':
        if computer_selection == 'Paper':
            print('User win')
            user_score += 1
        elif computer_selection == 'Rock':
            print('Computer win')
            computer_score += 1

print(f'User Score : {user_score}')
print(f'Computer Score : {computer_score}')
print('\n', '-' * 100, '\n')
```

# Task 6 
## chose to end 

```python
import random

while True:

    # Loop over 3 rounds
    user_score = 0
    compu_score = 0

    for _ in range(3):
        
        user_selection = input('Choose one of the following : rock, paper or scissor').capitalize()

        # Make a list of options for computer
        game_options = ['Rock', 'Paper', 'Scissor']

        # Remove the `user_selection` option
        game_options.remove(user_selection)

        # Computer selects one Option Randomlly
        computer_selection = random.choice(game_options)

        # Add Game Rules
        if user_selection == 'Rock':

            if computer_selection == 'Scissor':
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, User Wins!')
                user_score += 1         # user_score = user_score + 1

            else:
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, Computer Wins!')
                compu_score += 1


        elif user_selection == 'Paper':

            if computer_selection == 'Rock':
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, User Wins!')
                user_score += 1

            else:
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, Computer Wins!')
                compu_score += 1

        elif user_selection == 'Scissor':

            if computer_selection == 'Paper':
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, User Wins!')
                user_score += 1

            else:
                print(f'User selectes {user_selection}, Computer Selectes {computer_selection}, Computer Wins!')
                compu_score += 1


    print(f'User Score : {user_score}')
    print(f'Computer Score : {compu_score}')
    print('\n', '-' * 100, '\n')

    play_again = input('Would you like to play again ? [Y/N]').upper()

    if play_again == 'N':
        print('Thank you for playing Rock, Paper, Scissor :D')
        break
```
